## Partie 1 : Définitions et classification

### 1. Définitions

1. Un système temps réel est un système dont la correction dépend à la fois du résultat calculé et du moment auquel il est produit. Il doit respecter des contraintes temporelles.
2. Un résultat exact mais livré après son échéance peut être inutile ou dangereux : par exemple, une commande de freinage correcte mais trop tardive ne protège plus le véhicule.
3. **Période (`T`)** : intervalle entre deux activations successives d'une tâche périodique. **Temps d'exécution (`C`)** : durée processeur nécessaire à l'exécution d'une instance (on considère souvent son pire cas, WCET). **Échéance (`D`)** : délai maximal entre l'activation et la fin attendue. **Latence** : délai entre l'événement déclencheur et la réponse du système.
4. Une contrainte **stricte (temps réel dur)** ne tolère aucun dépassement d'échéance pertinent : le résultat tardif est considéré comme une défaillance. Une contrainte **non stricte** tolère certains retards : le résultat peut perdre toute valeur (temps réel ferme) ou seulement dégrader la qualité du service (temps réel mou).

### 2. Classification

| Système | Type : strict, ferme ou mou | Conséquence d'un retard |
|---|---|---|
| Radar de détection | Strict (si utilisé pour la sécurité) | Une détection tardive peut empêcher une réaction à temps et mettre des personnes en danger. |
| Système ABS | Strict | Une correction tardive du freinage peut entraîner une perte de contrôle ou allonger la distance d'arrêt. |
| Airbag | Strict | Un déclenchement après l'instant utile ne protège plus les occupants et peut aggraver les blessures. |
| Lecture audio/vidéo | Mou | Un retard provoque une coupure ou une image saccadée; les données tardives peuvent être abandonnées. |
| Supervision médicale | Ferme, ou strict pour une alarme vitale | Une mesure tardive peut devenir périmée; le retard d'une alarme critique peut mettre le patient en danger. |
| Contrôle d'un réacteur chimique | Strict | Une réaction tardive peut laisser dépasser les limites de sûreté et provoquer un accident grave. |

**Justifications :** L'ABS est strict, car la commande doit agir pendant le freinage; une réponse correcte mais tardive ne remplit plus sa fonction de sécurité. La lecture audio/vidéo est généralement molle : une image manquée dégrade la fluidité, mais le système peut continuer en abandonnant cette image.

## Partie 2 : Analyse fonctionnelle

| Système | Capteurs ou événements | Traitement | Actionneurs ou sortie | Échéance |
|---|---|---|---|---|
| Radar | Échos radio reçus par l'antenne | Filtrage, estimation de distance/vitesse, détection et suivi de cibles | Pistes affichées à l'opérateur, alerte ou commande associée | Avant que la cible n'atteigne une zone de risque; cycle déterminé par la portée et la vitesse |
| ABS | Vitesse de rotation des roues, pression de freinage | Détection du blocage et calcul de la correction de pression | Modulateurs hydrauliques des freins | Quelques millisecondes par cycle de contrôle, avant le blocage prolongé |
| Airbag | Accéléromètres et capteurs d'impact | Détection et classification du choc | Déclencheur pyrotechnique et gonflage du coussin | Quelques dizaines de millisecondes après le choc, avant le contact de l'occupant avec l'habitacle |
| Décodage audio/vidéo | Paquets compressés, horodatages et horloge de lecture | Décodage, synchronisation et préparation des échantillons/images | Haut-parleurs via DAC et images à l'écran | Avant l'instant de lecture; typiquement une échéance par trame ou bloc audio |
| Supervision médicale | ECG, pression artérielle, saturation ou autres mesures | Filtrage, calcul de paramètres et détection de seuils/anomalies | Affichage et alarme sonore/visuelle au personnel | Assez tôt pour permettre une intervention; délai fixé par le risque médical |
| Réacteur chimique | Température, pression, débit et niveau | Régulation, surveillance des limites et logique d'arrêt sûr | Vannes, pompes, chauffage/refroidissement et arrêt d'urgence | Avant qu'une variable ne dépasse sa limite de sûreté; dépend de la dynamique du procédé |

- Tâche 1 : `C₁ = 2 ms`, `T₁ = D₁ = 10 ms` ;

2. Utilisation totale : `U = U₁ + U₂ = 0,4`, soit **40 %**.
3. En une seconde, la tâche 1 est lancée `1000/10 = 100` fois et la tâche 2 `1000/20 = 50` fois.

5. La borne de Liu & Layland pour deux tâches vaut `2(√2 - 1) ≈ 0,8284` (82,84 %). Comme `U=0,4 ≤ 0,8284`, la condition suffisante est satisfaite.

## Partie 4 : Simulation Python

Modifiez les paramètres `C`, `T` et `duree_simulation` dans la cellule suivante. Observez les échéances respectées ou violées.

Testez au minimum les cas suivants :
- `C = 2`, `T = 5` ;
- `C = 5`, `T = 5` ;
- `C = 6`, `T = 5`.

Expliquez dans la cellule Markdown suivante à partir de quelle condition l'ordonnancement devient impossible.

```python
import matplotlib.pyplot as plt

# Paramètres à modifier
C = 2
T = 5
duree_simulation = 30

activations = list(range(0, duree_simulation, T))
fin_execution = [min(activation + C, duree_simulation) for activation in activations]

fig, ax = plt.subplots(figsize=(11, 2.8))
for numero, (activation, fin) in enumerate(zip(activations, fin_execution), 1):
    ax.barh(0.25, fin - activation, left=activation, height=0.2, color="#2f80ed")
    ax.text((activation + fin) / 2, 0.25, f"Instance {numero}", ha="center", va="center", color="white", fontsize=8)
    ax.annotate("", xy=(activation + T, 0.02), xytext=(activation + T, 0.35), arrowprops={"arrowstyle": "<->", "color": "#d64545", "linestyle": "--"})

ax.spines["bottom"].set_position(("data", 0))
ax.spines["top"].set_visible(False)
ax.set_xlim(0, duree_simulation)
ax.set_ylim(-0.1, 0.7)
ax.set_yticks([0.25])
ax.set_yticklabels(["Processeur"])
ax.set_xticks(range(0, duree_simulation + 1, T))
ax.set_xlabel("Temps")
ax.set_title(f"Tâche périodique : C={C}, T={T}")
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()
```

### Interprétation de la simulation

1. Avec une échéance relative `D=T`, chaque instance respecte son échéance si `C≤T` dans ce modèle idéal. Si `C<T`, elle termine avant l'activation suivante et le processeur dispose de `T-C` unités de temps libres par période.
2. Lorsque `C>T`, une instance n'a pas terminé à l'arrivée de la suivante. La charge dépasse la capacité du processeur (`U=C/T>1`), le travail s'accumule et des échéances sont inévitablement manquées.
3. Lorsque `C=T`, chaque instance occupe toute sa période et finit exactement à son échéance. Il n'y a aucun temps libre pour les autres tâches, les interruptions ou les surcoûts; la moindre surcharge peut provoquer un retard.

## Synthèse finale

Un système temps réel doit produire des résultats corrects avant leurs échéances : un résultat exact livré trop tard peut être inutile ou dangereux. La conception doit donc tenir compte des périodes, des temps d'exécution et des échéances, puis vérifier que la charge totale reste compatible avec la capacité du processeur et que les tâches critiques respectent leur temps de réponse. Enfin, le comportement en surcharge doit être prévu selon les conséquences d'un retard : abandonner des données périmées peut convenir au multimédia, mais les fonctions de sécurité doivent rester prioritaires et mener si nécessaire à un état sûr.
